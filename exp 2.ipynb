```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

df = pd.read_csv('/Users/vicram/Downloads/student-por.csv')

df.head()
```

```python
df.info()
df.describe()
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
print("Average G1:", df['G1'].mean())
print("Average G2:", df['G2'].mean())
print("Average G3:", df['G3'].mean())
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['G3'],
    bins=10,
    kde=True
)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=df['G3']
)

plt.title('Final Grade Distribution')
plt.xlabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='studytime',
    y='G3'
)

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='absences',
    y='G3'
)

plt.title('Absences vs Final Grade')
plt.xlabel('Number of Absences')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='sex',
    y='G3'
)

plt.title('Gender vs Final Grade')
plt.xlabel('Gender')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='internet',
    y='G3'
)

plt.title('Internet Access vs Final Grade')
plt.xlabel('Internet Access')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='G1',
    y='G3'
)

plt.title('First Period Grade vs Final Grade')
plt.xlabel('G1 Grade')
plt.ylabel('G3 Final Grade')

plt.show()
```
