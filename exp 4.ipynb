```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

df = pd.read_csv('/Users/vicram/Downloads/student-por.csv')

print(df.head())
```

```python
print(df['sex'].value_counts())

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='sex'
)

plt.title('Number of Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Students')

plt.show()
```

```python
print(df['G3'].describe())

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='G3',
    bins=10,
    kde=True
)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Number of Students')

plt.show()
```

```python
grade_data = df[['G1', 'G2', 'G3']].mean()

print(grade_data)

plt.figure(figsize=(8,5))

grade_data.plot(
    kind='bar'
)

plt.title('Average Student Grades')
plt.xlabel('Grade')
plt.ylabel('Average Marks')

plt.xticks(
    rotation=0
)

plt.show()
```

```python
print(df['studytime'].value_counts().sort_index())

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='studytime'
)

plt.title('Distribution of Study Time')
plt.xlabel('Study Time')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='studytime',
    y='G3'
)

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')

plt.show()
```

```python
print(df['absences'].describe())

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='absences',
    bins=20,
    kde=True
)

plt.title('Distribution of Student Absences')
plt.xlabel('Number of Absences')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='absences',
    y='G3'
)

plt.title('Absences vs Final Grade')
plt.xlabel('Number of Absences')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='sex',
    y='G3'
)

plt.title('Gender vs Final Grade')
plt.xlabel('Gender')
plt.ylabel('Final Grade')

plt.show()
```

```python
print(df['internet'].value_counts())

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='internet',
    y='G3'
)

plt.title('Internet Access vs Final Grade')
plt.xlabel('Internet Access')
plt.ylabel('Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='G1',
    y='G3'
)

plt.title('First Period Grade vs Final Grade')
plt.xlabel('G1 Grade')
plt.ylabel('G3 Final Grade')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='G2',
    y='G3'
)

plt.title('Second Period Grade vs Final Grade')
plt.xlabel('G2 Grade')
plt.ylabel('G3 Final Grade')

plt.show()
```

```python
correlation = df[
    ['age', 'studytime', 'failures', 'absences', 'G1', 'G2', 'G3']
].corr()

print(correlation)

plt.figure(figsize=(9,7))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Matrix of Student Performance')

plt.show()
```

```python
gender_average = df.groupby('sex')['G3'].mean()

print(gender_average)

plt.figure(figsize=(7,5))

gender_average.plot(
    kind='bar'
)

plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average G3')

plt.xticks(
    rotation=0
)

plt.show()
```

```python
study_average = df.groupby('studytime')['G3'].mean()

print(study_average)

plt.figure(figsize=(8,5))

study_average.plot(
    kind='bar'
)

plt.title('Average Final Grade by Study Time')
plt.xlabel('Study Time')
plt.ylabel('Average G3')

plt.xticks(
    rotation=0
)

plt.show()
```

```python
top_students = df.sort_values(
    by='G3',
    ascending=False
).head(10)

print(
    top_students[
        ['school', 'sex', 'age', 'studytime', 'G1', 'G2', 'G3']
    ]
)
```
